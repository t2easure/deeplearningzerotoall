## convolution
이미지 위에서 stride 값 만큼 filter를 이동시키면서 겹쳐지는 부분의 각 원소의 값을 곱해 모두 더한 값을 출력으로 하는 연산

- stride : filter를 한 번에 얼마나 이동할 것인가
- padding : zero-padding -> padding이 1인 경우
    

### pytorch nn.Conv2d
class 선언하는 법
- torch.nn.Conv2d(in_channels, out_channels, kernel_size, stride=1, padding=0, bias=True...)
- 입력의 형태
1. input type : torch.Tensor
2. input shape : (N x C x H x W) = (batch_size, channel, height, width)

### convolution의 output 크기
output size = (input size - filter size + 2 * padding) / stride + 1

## 뉴런과 convolution 관계

* **한 줄 요약:**
    * Convolution(합성곱)은 **"네모난 모양을 가진 퍼셉트론(뉴런)"**이다.

* **구성 요소 매칭:**
    1.  **입력 ($x$):** 이미지의 특정 조각 (Receptive Field)
    2.  **가중치 ($w$):** 필터 (커널)
    3.  **편향 ($b$):** Bias

* **연산 원리:**
    * 기존 뉴런과 똑같이 **[입력 $\times$ 가중치]**를 전부 더하는 **내적(Weighted Sum)** 계산을 수행함.
    * $$y = \sum (x \cdot w) + bias$$

## pooling
이미지 사이즈를 줄이기 위한 연산
- max pooling : size 이내에서 가장 큰 값만 남김
- average pooling : size 이내 모든 값 평균 남김

### maxpool2d
- class : torch.nn.MaxPool2d(kernel_size, stride=None, padding=0)
