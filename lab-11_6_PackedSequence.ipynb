## PackedSequence와 PaddedSequence
RNN / LSTM 계열의 모델에서 sequence batch를 잘 활용할 수 있는 법
1. 병렬 연산 효율성(padding) : 서로 다른 길이의 문장들을 배치 단위로 묶기 위해, 가장 긴 문장에 맞춰 빈 공간을 0<PAD>로 채워 직사각형 행렬 모양 만듦
2. 연산 낭비 방지 (Packing) : RNN이 의미 없는 패딩(0) 부분을 연산하지 않고, 실제 단어들만 골라 학습하게 해 처리 속도 높임
3. Hidden state 정확도 : 문장의 실제 끝 지점을 정확히 인식하게 함으로써, 마지막 hidden state에 패딩 노이즈가 섞이지 않도록 결과의 정확성 보장

- PyTorch가 제공하는 함수들
1. pad_sequence
2. pack_sequence
3. pack_padded_sequence
4. pad_packed_sequence

```python
import torch 
import numpy as np
from torch.nn.utils.rnn import pad_sequence, pack_sequence, pack_padded_sequence, pad_packed_sequence
```
